```python
import pandas as pd

def adatok():
  url = (
      'https://raw.githubusercontent.com/plotly/datasets/master/imports-85.csv'
  )
  df = pd.read_csv(url, na_values='?')
  return df
    
# 1. kerdes: leggyakoribb markak
def kerdes1(df):
  print('1. Leggyakoribb markak az adatsorban:')
  print(df['make'].value_counts().head(3))
  print()

# 2. kerdes: uzemanyag es atlagar
def kerdes2(df):
  print('2. Uzemanyag tipusok atlagara:')
  print(df.groupby('fuel-type')['price'].mean())
  print()

# 3. kerdes: meghajtas es loero
def kerdes3(df):
  print('3. Meghajtas es atlagos loero:')
  print(df.groupby('drive-wheels')['horsepower'].mean())
  print()

# 4. kerdes: fogyasztas es ar korrelacioja
def kerdes4(df):
  print('4. Varosi fogyasztas es ar kapcsolata (korrelacio):')
  korr = df['city-mpg'].corr(df['price'])
  print(korr)
  print()

# 5. kerdes: draga autok szurese
def kerdes5(df):
  print('5. Kiugroan draga autok listaja:')
  atlag = df['price'].mean()
  szoras = df['price'].std()
  hatar = atlag + (2 * szoras)

  draga = df[df['price'] > hatar]
  print(draga[['make', 'price']])

df = adatok()
kerdes1(df)
kerdes2(df)
kerdes3(df)
kerdes4(df)
kerdes5(df)
```
